# Clustering Analysis – Wine Quality Dataset

## 1. Introduction
This notebook applies K-Means clustering to the Wine Quality dataset. The goal is to identify natural groupings of wines based on their physicochemical properties. Proper preprocessing steps such as feature scaling and dimensionality reduction are applied to ensure valid clustering results.

## 2. Data Loading

```
import pandas as pd

df = pd.read_csv('winequality-red.csv', sep=';')
df.head()
```

## 3. Feature Selection

All physicochemical variables are used for clustering, excluding the target variable `quality`. This avoids arbitrary feature selection and preserves the multivariate structure of the data.

```
X = df.drop('quality', axis=1)
```

## 4. Feature Scaling

K-Means is a distance-based algorithm, so feature scaling is required to prevent variables with larger numeric ranges from dominating the clustering process.

```
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 5. Elbow Method for Optimal Clusters

```
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

The elbow in the plot suggests that **k = 3** provides a reasonable balance between cluster compactness and model simplicity.

## 6. K-Means Clustering

```
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

df['Cluster'] = clusters
```

## 7. Cluster Interpretation

To interpret the clusters, the mean values of each physicochemical variable are examined for each cluster.

```
df.groupby('Cluster').mean()
```

## 8. Dimensionality Reduction for Visualization

Principal Component Analysis (PCA) is used to project the high-dimensional data into two dimensions for visualization purposes.

```
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

## 9. Cluster Visualization

```
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Wine Clusters (PCA Projection)')
plt.show()
```

## 10. Conclusion
K-Means clustering was successfully applied to the Wine Quality dataset using appropriate preprocessing steps. Feature scaling ensured valid distance calculations, and PCA enabled meaningful visualization. The resulting clusters reveal distinct groupings of wines based on their chemical characteristics, providing useful insights for further analysis.
